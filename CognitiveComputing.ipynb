```python
import random
random.seed(1024170334)

numbers = [random.randint(100,900) for _ in range(100)]
print(numbers)

cntodd=0
for num in numbers:
    if num%2 != 0:
        print(num, end=" ")
        cntodd += 1
print("total odds are: ", end=" ")
print(cntodd)

cnteven=0
for num in numbers:
    if num%2 == 0:
        print(num, end=" ")
        cnteven += 1
print("total evens are: ", end=" ")
print(cnteven)


def isprime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n%i == 0:
            return False
    return True

primeNumbers = [num for num in numbers if isprime(num)]

print("prime numbers are: ")
print(primeNumbers)
print("Total prime numbers are: ", len(primeNumbers));

mostfreq = max(numbers, key=numbers.count)

count = numbers.count(mostfreq)
print("most frequent number is: ",mostfreq," frequency of number is: ",count)
```

```python

rollno = 1024170315
A=set()
B = set()

temp=rollno

while temp>0:
    d = temp%10
    A.add(d*7)
    B.add(d*9)
    temp = temp//10
print(A)
print(B)

print(A.union(B))
print(A.intersection(B))

print(A.difference(B))
print(B.difference(A))

print(A.symmetric_difference(B))
print(A.issubset(B))
print(A.issuperset(B))
print(B.issubset(A))
print(B.issuperset(A))

val=int(input("enter value to remove"))
A.discard(val)# discard does not raise a key error like remove does
print(A)
```

```python
my_dict = {
    "name": "Rohit Poddar",
    "roll_no": "1024170315",
    "branch": "COPC",
    "age": 21,
    "city": "Patiala"
}

my_dict["location"] = my_dict.pop("city")
# my_dict.insert("CGPA")
my_dict["CGPA"]= "7.98"
my_dict["age"] = my_dict["age"] + 1
print(my_dict)

dict1 = my_dict.copy()
dict2 = my_dict.copy()

rmvvalue = dict1.pop("branch") #pop key removes and return the value
print(rmvvalue)

del dict2["branch"] # del only remove donot return the value
print(dict2)

for key,value in my_dict.items():
    print(f"{key} = {value}")  #using f is important (format)

if "email" in my_dict:
    print("email:",my_dict["email"])
else:
    print("Email key not found in dictionary")


new_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024170999",
    "branch": "COE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **new_dict}  # the SECOND dictionary value overwrites FIRST

print("My Dictionary:")
print(my_dict)

print("\nFriend Dictionary:")
print(new_dict)

print("\n")
print(merged_dict)
      

string_dict = {key: value for key,value in my_dict.items() if isinstance(value, str)}

print(my_dict)
print(string_dict)
```
