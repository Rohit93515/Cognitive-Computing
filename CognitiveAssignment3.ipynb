```python
import pandas as pd

emp_data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv('employees.csv', index=False)
print("employees.csv created:\n", emp_df, "\n")

emp_df = pd.read_csv('employees.csv')

print("Shape (rows, columns):", emp_df.shape, "\n")

print("DataFrame info:")
emp_df.info()
print()

print("Descriptive statistics:\n", emp_df.describe(), "\n")

print("First 5 rows:\n", emp_df.head(), "\n")
print("Last 3 rows:\n", emp_df.tail(3), "\n")

avg_salary = emp_df['Salary'].mean()
total_bonus = emp_df['Bonus'].sum()
youngest_age = emp_df['Age'].min()
highest_rating = emp_df['Rating'].max()

print("Average salary:", avg_salary)
print("Total bonus paid:", total_bonus)
print("Youngest employee's age:", youngest_age)
print("Highest performance rating:", highest_rating, "\n")

sorted_by_salary = emp_df.sort_values(by='Salary', ascending=False)
print("Sorted by Salary (descending):\n", sorted_by_salary, "\n")

def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

emp_df['Performance_Category'] = emp_df['Rating'].apply(categorize_rating)
print("DataFrame with Performance_Category:\n", emp_df, "\n")

print("Missing values per column:\n", emp_df.isnull().sum(), "\n")

emp_df = emp_df.rename(columns={'Employee_ID': 'ID'})
print("Columns after rename:", list(emp_df.columns), "\n")

more_than_5_years = emp_df[emp_df['Years_of_Experience'] > 5]
print("Employees with more than 5 years experience:\n", more_than_5_years, "\n")

it_department = emp_df[emp_df['Department'] == 'IT']
print("Employees in IT department:\n", it_department, "\n")

emp_df['Tax'] = emp_df['Salary'] * 0.10
print("DataFrame with Tax column:\n", emp_df, "\n")

emp_df.to_csv('employees_modified.csv', index=False)
print("Modified DataFrame saved to 'employees_modified.csv'")
```
