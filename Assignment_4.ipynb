```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])
```

```python
import numpy as np
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

Rohit = np.linspace(10, 100, 25)

print("Array:", Rohit)

print("Dimensions:", Rohit.ndim)

print("Shape:", Rohit.shape)

print("Total elements:", Rohit.size)

print("Data type:", Rohit.dtype)

print("Total bytes:", Rohit.nbytes)

transpose_array = Rohit.reshape(1, 25).T

print("Transpose using reshape():")
print(transpose_array)
print("Transpose using T attribute:")
print(Rohit.T)
```

```python
import numpy as np

ucs420_rohit = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_rohit)

print("\nMean:", np.mean(ucs420_rohit))
print("Median:", np.median(ucs420_rohit))
print("Maximum:", np.max(ucs420_rohit))
print("Minimum:", np.min(ucs420_rohit))
print("Unique elements:", np.unique(ucs420_rohit))

reshaped_ucs420_rohit = ucs420_rohit.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_rohit)

resized_ucs420_rohit = np.resize(ucs420_rohit, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_rohit)
```
